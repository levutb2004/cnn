```python
import rasterio
import numpy as np


root_path = "E:/pop-disag/fake-pomelo/pomelo_input_data/"

input_paths = {
    "vnm": {
        "buildings_google": f"{root_path}VNM/covariates/building_counts_2019_100m.tif",
        "buildings_maxar": f"{root_path}VNM/covariates/building_counts_2019_100m.tif",
        "buildings_google_mean_area": f"{root_path}VNM/covariates/vnm_built_S_GHS_U_wFGW_100m_v1_2019.tif",
        "buildings_maxar_mean_area": f"{root_path}VNM/covariates/vnm_built_S_GHS_U_wFGW_100m_v1_2019.tif",
        "built_s": f"{root_path}VNM/covariates/vnm_built_S_GHS_U_wFGW_100m_v1_2019.tif",
        "built_v": f"{root_path}VNM/covariates/vnm_built_V_GHS_U_wFGW_100m_v1_2019.tif",
        "vnm_nightlights": f"{root_path}VNM/covariates/vnm_viirs_nvf_2019_100m_v1.tif",
        "vnm_elevation": f"{root_path}VNM/covariates/vnm_elevation_merit103_100m_v1.tif",
        "vnm_highway_dist": f"{root_path}VNM/covariates/vnm_highway_dist_osm_2023_100m_v1.tif",
        "vnm_road_intr_dist": f"{root_path}VNM/covariates/vnm_rd_intrs_dist_osm_2023_100m_v1.tif",
        "vnm_waterbodies_dist": f"{root_path}VNM/covariates/vnm_waterbodies_dist_osm_2023_100m_v1.tif",
        "vnm_inland_water_dist": f"{root_path}VNM/covariates/vnm_dist_inland_water_100m_esa_2021_v1.tif",
        "vnm_coastline_dist": f"{root_path}VNM/covariates/vnm_coastline_dst_100m_v1.tif",
        "vnm_wdpa_dist": f"{root_path}VNM/covariates/vnm_WDPA_pre2019_cat1_dist_100m_v1.tif",
        "vnm_esalc_11": f"{root_path}VNM/covariates/vnm_esalc_11_dst_2019_100m_v1.tif",
        "vnm_esalc_40": f"{root_path}VNM/covariates/vnm_esalc_40_dst_2019_100m_v1.tif",
        "vnm_esalc_130": f"{root_path}VNM/covariates/vnm_esalc_130_dst_2019_100m_v1.tif",
        "vnm_esalc_140": f"{root_path}VNM/covariates/vnm_esalc_140_dst_2019_100m_v1.tif",
        "vnm_esalc_150": f"{root_path}VNM/covariates/vnm_esalc_150_dst_2019_100m_v1.tif",
        "vnm_esalc_160": f"{root_path}VNM/covariates/vnm_esalc_160_dst_2019_100m_v1.tif",
        "vnm_esalc_190": f"{root_path}VNM/covariates/vnm_esalc_190_dst_2019_100m_v1.tif",
        "vnm_esalc_200": f"{root_path}VNM/covariates/vnm_esalc_200_dst_2019_100m_v1.tif",
        "vnm_bsgme": f"{root_path}VNM/covariates/vnm_bsgme_v0a_100m_2019_resampled.tif",
        "vnm_bsgme_dst": f"{root_path}VNM/covariates/vnm_dst_bsgme_100m_2019_resampled.tif",
        "slope": f"{root_path}VNM/covariates/vnm_slope_merit103_100m_v1.tif",
    }
}


norms = {}

print("=" * 80)
print("RASTER NODATA + NORMALIZATION STATISTICS")
print("=" * 80)

for name, path in input_paths["vnm"].items():

    with rasterio.open(path) as src:

        data = src.read(1).astype(np.float64)

        # Nodata lấy TRỰC TIẾP từ raster
        nodata = src.nodata

        print(f"\n{name}")
        print(f"  nodata = {nodata}")

        # Valid pixels
        valid_mask = np.isfinite(data)

        if nodata is not None:
            if nodata > 1e30:
                valid_mask &= ~np.isclose(data, nodata)
            else:
                valid_mask &= data != nodata

        values = data[valid_mask]

        mean = values.mean()
        std = values.std()

        print(f"  mean   = {mean}")
        print(f"  std    = {std}")

        norms[name] = (mean, std)


# ============================================================
# OUTPUT FOR configpop.py
# ============================================================

print("\n\n")
print("=" * 80)
print("COPY TO configpop.py")
print("=" * 80)

print("\nnorms = {")
print('    "vnm": {')

for name, (mean, std) in norms.items():
    print(
        f'        "{name}": '
        f'({mean:.15g}, {std:.15g}),'
    )

print("    },")
print("}")
```
